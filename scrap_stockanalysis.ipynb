```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
# URL to scrape
base_url = "https://stockanalysis.com/"
url = base_url+"/markets/gainers/"

columns = ['name','change', 'source']
df = pd.DataFrame(columns=columns)

# Send a GET request to the URL
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url, headers=headers)

# Check if request was successful
if response.status_code != 200:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
    exit()

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")


gainers_table = soup.find('table', {'id': 'main-table'})
table_rows = gainers_table.find_all('tr')
print(len(table_rows))            #.find_all("div",{'class':['feed-row', 'feed-border-gradient', 'rounded mb-3']})


for i in range(1,7):
    print("I -> ",i)
    cols_in_row = table_rows[i].find_all('td')
    name = cols_in_row[1].get_text()
    # link = cols_in_row[1].find('a').get('href')

    change = cols_in_row[3].get_text()
    change_num = float(re.sub(r'[+%]', '', change))
    

    print('stock_name: ',name)
    print('change: ',change_num)

    new_row = {
    'name': name,
    'change': change_num,
    'source': 'StockAnalysis'
    }

    df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    df.to_csv("stockanalysis_name_change.csv")

```

```python
# URL to scrape
url = "https://www.stocktitan.net/scanner/momentum"

# Send a GET request to the URL
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url, headers=headers)

# Check if request was successful
if response.status_code != 200:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
    exit()

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")

# Find the TOP GAINERS table
# It is identified by the heading just before it
h2_tags = soup.find_all("h2")
top_gainers_table = None
for h2 in h2_tags:
    if "Top Gainers" in h2.text:
        # The next sibling after the <h2> should be the table
        next_sibling = h2.find_next_sibling("div")
        if next_sibling and next_sibling.find("table"):
            top_gainers_table = next_sibling.find("table")
            break

# Ensure the table was found
if not top_gainers_table:
    print("TOP GAINERS table not found.")
    exit()

# Extract headers
headers = [th.text.strip() for th in top_gainers_table.find_all("th")]

# Extract rows
rows = []
for tr in top_gainers_table.find_all("tr")[1:]:  # skip header row
    cells = [td.text.strip() for td in tr.find_all("td")]
    if cells:
        rows.append(cells)

# Convert to DataFrame and display
df = pd.DataFrame(rows, columns=headers)
print(df.head())
```
